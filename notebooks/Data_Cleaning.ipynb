```python
# Detect outliers using IQR

numerical_columns = [
    "Quantity",
    "UnitPrice",
    "ItemsInCart",
    "TotalPrice"
]

iqr_summary = []

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)
    upper = Q3 + (1.5 * IQR)

    outliers = df[(df[column] < lower) | (df[column] > upper)]

    iqr_summary.append({
        "Column": column,
        "Outliers": len(outliers),
        "Lower Bound": round(lower,2),
        "Upper Bound": round(upper,2)
    })

iqr_summary = pd.DataFrame(iqr_summary)

iqr_summary
```

```python
# Detect Outliers Using the Z-Score Method

from scipy.stats import zscore

zscore_summary = []

for column in numerical_columns:

    z_scores = zscore(df[column])

    outliers = df[abs(z_scores) > 3]

    zscore_summary.append({
        "Column": column,
        "Outliers": len(outliers)
    })

zscore_summary = pd.DataFrame(zscore_summary)

zscore_summary
```

```python
# Visualise Using Boxplots

import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

df[numerical_columns].boxplot()

plt.title("Boxplots Before Outlier Treatment")

plt.show()
```

```python
# Visualise Using Scatterplots

for column in numerical_columns:

    plt.figure(figsize=(7,4))

    plt.scatter(df.index,
                df[column])

    plt.title(f"Scatterplot of {column}")

    plt.xlabel("Observation")

    plt.ylabel(column)

    plt.show()
```

```python
# Apply IQR Capping

# Create a copy

df_clean = df.copy()

for column in numerical_columns:

    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)
    upper = Q3 + (1.5 * IQR)

    df_clean[column] = df_clean[column].clip(lower, upper)
```

```python
# Run the IQR detection again

iqr_after = []

for column in numerical_columns:

    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR

    outliers = df_clean[
        (df_clean[column] < lower) |
        (df_clean[column] > upper)
    ]

    iqr_after.append({
        "Column": column,
        "Remaining Outliers": len(outliers)
    })

pd.DataFrame(iqr_after)
```

```python
# Boxplots After Treatment

plt.figure(figsize=(12,6))

df_clean[numerical_columns].boxplot()

plt.title("Boxplots After Outlier Treatment")

plt.show()
```
